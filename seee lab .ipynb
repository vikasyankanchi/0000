```python
import spacy
from nltk import ngrams
nlp = spacy.load('en_core_web_sm')
doc = nlp("AI is simulation of human intelligence by machine. it includes expects system nlp speech recognition and machin vision")
bigrams = ngrams(doc,2)
print(list(bigrams))
```
